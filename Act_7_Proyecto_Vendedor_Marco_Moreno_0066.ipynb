```python
# =========================================
# IA PARA VENDEDORES Y CLIENTES
# Alumno: Espinoza Moreno Marco Antonio
# Especialidad: Inteligencia Artificial
# Maestro: Eliseo Nava
# Plantel: CBTis 128
# Número de lista: 16
# =========================================

print("===================================")
print("   SISTEMA IA PARA VENDEDORES")
print("===================================")

nombre = input("Ingresa tu nombre: ")

# MENU PRINCIPAL
while True:

    print("\n========== MENU ==========")
    print("1. Revisar ventas")
    print("2. Revisar comisiones")
    print("3. Evaluación de vendedor")
    print("4. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCION 1
    if opcion == "1":

        ventas = int(input("¿Cuántas ventas realizaste hoy?: "))

        if ventas >= 10:
            print("IA:", nombre, "Excelente rendimiento en ventas.")

        elif ventas >= 5:
            print("IA:", nombre, "Buen trabajo, vas mejorando.")

        else:
            print("IA:", nombre, "Necesitas mejorar tus ventas.")

    # OPCION 2
    elif opcion == "2":

        monto = float(input("Ingresa el monto total vendido: "))

        if monto >= 10000:
            print("IA:", nombre, "Ganaste comisión alta.")

        elif monto >= 5000:
            print("IA:", nombre, "Ganaste comisión media.")

        else:
            print("IA:", nombre, "Ganaste comisión básica.")

    # OPCION 3
    elif opcion == "3":

        actitud = input("¿Llegaste puntual? (si/no): ")

        if actitud == "si":
            print("IA:", nombre, "Tienes buena disciplina laboral.")

        else:
            print("IA:", nombre, "Debes mejorar tu puntualidad.")

    # OPCION 4
    elif opcion == "4":

        print("Gracias por usar la IA para vendedores.")
        break

    # OPCION INVALIDA
    else:

        print("Opción no válida.")
        print("Programa desarrollado por Espinoza Moreno Marco Antonio")
```
